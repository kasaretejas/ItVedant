```python
# Data types in python
# data types are used to categories what kind of data variable holds
# Numeric    :  int --> whole numbers, float --> decimal numbers, complex ---> a+bj, here a and b can be int/float
# characters :  str (string)
# yes/no     :  boolean (true/false)
# sequence   : list [12, 56, 88, 12] ---> allowed duplicate + mutable, tuple (12, 56, 88, 12) --> allowed duplicate + immutable
# sequense   : set {12, 56, 88} ---> unique values + mutable, frozenset {12, 56, 88} --->unique values + immutable
# key-values : dict (dictionary) {"name":"raj", "age":24}
```

```python
p = 10
print(p)
print(type(p)) # what is data type of p

p = 3.8
print(p)
print(type(p)) # what is data type of p
```

```python
print (12, type(12))
print (3.6, type(3.6))
print (6+9j, type(6+9j))

print ("hello", type("hello"))
print ("aR2$", type("aR2$"))
print ('aR2$', type('aR2$'))
print ("""python""", type("""python"""))  #for multi line string
print ('''python''', type('''python'''))

print(True , type(True))
```

```python
print ([10,20,10], type([10,20,10]))
print ((10,20,10), type((10,20,10)))
print ({10,20,10}, type({10,20,10}))
print (frozenset({10,20,10}), type(frozenset({10,20,10})))

print ({"name":"raj", "age":24}), type({"name":"raj", "age":24})

student = {"name":"raj", "age":24}
print(student)
print(type(student))
```
