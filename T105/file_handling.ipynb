```python
#basic modes
# w : write ----> override existing data
# r: read ------> to read
# a: append ----> dosnt override existing data, appends new data at the end of existing data
# x: -----------> to create new file

# how to open file
# 1. variable = open("file_name", "mode")  -----> simplest way but we have to close file manually

# 2. with open("file_name", "mode") as variable: ----> best way for auto close
```

```python
# creating a file
print("file about to create")
file=open("sample.txt","x")
print("file sample.txt is created")
file.close()
print("file sample.txt is closed")
```

```python
# how to check file is closed manually
file = open("sample.txt", "r")
print("does file is closed? ", file.closed)
file.close()
print("does file is closed? ", file.closed)
```

```python
# file closing manually using with
with open("sample.txt", "r") as file:
    print("reading data from file.........")
    print("does file is closed? ", file.closed)

print("does file is closed? ", file.closed)
```

```python
# writting data into file
with open("sample.txt", "w") as file:
    file.write("hello\n")
    file.write("python")
print("data writting done!")
```

```python
# reading data from file ---> read() 
with open("sample.txt", "r") as file:
    fileContent=file.read() 
    print(fileContent)
print("data reading done")
```

```python
# reading data from file ---> readline()
with open("sample.txt", "r") as file:
    fileContent=file.readline()
    print(fileContent)
print("data reading done")
```

```python
# reading data from file
with open("sample.txt", "r") as file:
    fileContent=file.readlines()
    print(fileContent)
print("data reading done")
```

```python
# create a new file and add/write table of 2 in it

with open("demo.txt", "w") as file:    # if you open file using with then that file will be created automatically
    for num in range(1,11):
        file.write(str(2*num)+"\n")
print("table of 2 is added into file")
```

```python
# ask user his/her name and write that name into person.txt file
file=open("person.txt","x")
file.close()
print("file is created and closed")
```

```python
with open("person.txt","a") as file:
    name = input("enter your name")
    file.write(name+"\n")
print("name added to file")
```
