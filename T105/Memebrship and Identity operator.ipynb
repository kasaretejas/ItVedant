```python
# memebership operator :  to check member in a given sequence ==> in, not in
print("z" in "zebra")
print("p" not in "zebra")
print("Z" in "zebra")

name = "amit"
print("t" in name)
print("y" in name)
print("i" not in name)


marks = [90, 97,78, 45]
print(90 in marks)
print(80 not in marks)
```

```python
#identity operator : check identity/memory location of given objects/variable ===> is, is not
```

```python
a = 10
b = 10
print(a == b)
print(a is b)

x = 257    #in python, for int,  same memory location is shared in range -5 to 256
y = 257
print(x == y)
print(x is y)


p = 1.5
q = 1.5
print(p == q)
print(p is q)
```

```python
# is, is not =====> compare memory location
# ==         =====> content/value comparision
```
