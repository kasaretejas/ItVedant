```python
# display all marks one by one --> 89,92,43,96
# display only even marks -------> 92,96
marks = [89,92,43,96]
for mark in marks:
    print(mark)
    
print("-----------------------")

for mark in marks:
    if(mark%2==0):
        print(mark)
```

```python
# if marks are less than 35, display as failed otherwise display pass.
# ex : 89 pass
marks = [89,92,13,96,31,76]
for mark in marks:
    if(mark<35):
        print(mark, "failed")
    else:
        print(mark, "pass")
```

```python
#break and continue : transfer statements
# because break and continue transfers control out of the loop OR towards the loop

# break : when we have to stop/terminate execution of loop based on condition
# continue : when we have to skip current iteration based on condition

prices = [199,99,299,199,99,399]
# do not buy item worth 99
for price in prices:
    if(price==99):
        continue #do not proceed, go toward next item/iteration
    else:
        print(price, "added to the cart")

print("-----------------------------------------")

# do not buy any item as soon as you get item worth 99
for price in prices:
    if(price==99):
        break #do not proceed, stop the loop
    else:
        print(price, "added to cart")
    
```

```python
#take passowrd from user until he/she enters correct passowrd
count= 0 
actualPassword = "abcd123"
enteredPassword=""
while(actualPassword != enteredPassword):
    enteredPassword = input("enter password")
    count+=1
    if(actualPassword == enteredPassword):
        print("Welcome!!!")
        break
    
print("you have entered wrong password ", count, " times")
```

```python
# guess a number
actualNumber = 56
guessedNumber = 0
while(actualNumber != guessedNumber):
    guessedNumber = int(input("guess a number"))
    if(guessedNumber < actualNumber):
        print("guess a number greater than", guessedNumber)
        continue
    elif(guessedNumber > actualNumber):
        print("guess a number less than", guessedNumber)
        continue
    else:
        print("Correct guess !!")
        break
```

```python
# while loop with alredy known termination
# display 1 to 5 numbers using while loop
n=1
while(n<=5):
    print(n)
    n+=1
```
