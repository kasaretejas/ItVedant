```python
print(12+5)
print(12-5)
print(12*5)
print(12/5)
print(12/6)
print(12%5)
print(12%3)

print("hii"+"bye")
```

```python
x = 10
print(x)
x = x+5 #10+5 = 15     ====> x+=5
print(x)
x-=8 #x=x-8 ==> 15-8 = 7
print(x)
```

```python
print(5>3)
print(5<3)
print(5>5)
print(5>=5)
print(5<5)
print(5<=5)
print(5!=5)
print(5==5)

print("hii"=="hii")
print("hii"=="hi")
print("ab">"zurich") #ascii code comparision a=97, A=65
print("ab">"Zurich")
```

```python
# and gives True if all conditions are True
print(8>3 and 8>5 and 8>1)
print(8>3 and 8>5 and 8<1)
# or gives True if at least one condition is True
print(8>3 or 8>5 or 8>1)
print(8<3 or 8<5 or 8>1)
print(8<3 or 8<5 or 8<1)
# not reverse the result
print(8>2)
print(not 8>2)
```
