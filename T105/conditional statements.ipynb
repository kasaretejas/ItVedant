```python
# problem : take age from user and check for voting eligibility
age = int(input("enter your age"))
if(age>=18):
    print("Eligible for voting")
else:
    print("Not eligible for voting")
```

```python
# problem : take number from user and check for even or odd
num = int(input("enter a number"))
if(num%2==0):
    print(num, " is even")
else:
    print(num, " is odd")
```

```python
# problem : take number from user and check divisible by 3 and must be even
# 15 is divisible by 3 but not even
# 12 is divisible by 3 and even
num = int(input("enter a number"))
if(num%3==0):
    if(num%2==0):
        print(num , " is correct number")
    else:
        print(num, " is divisible by 3 but not even")
else:
    print(num, " is not divisible by 3")
```

```python
#take number from user and check whether number is positive, negative OR zero
num = int(input("enter a number "))
if(num>0):
    print(num, " is positive")
elif(num<0):
    print(num, " is negative")
else:
    print(num, " is zero")
```

```python
# take 3 numbers from user and tell which is largest number
x=int(input("enter first number"))
y=int(input("enter second number"))
z=int(input("enter third number"))

if(x>y and x>z):
    print(x, "is largest")
elif(y>x and y>z):
    print(y, "is largest")
else:
    print(z, "is largest")
```
