# Difference between list , tuple and set
marks = [90,89,45]  #list
ages  = (23, 18, 65) #tuple
cities= {"pune", "mumbai"} #set

1. list have [], tuple have (), set have {}
2. in list and tuple, duplicate elements are allowed but not in set
3. indexing is applicable only on list and tuple, not on set
4. if we insert elements in list or tuple, the insertion order is preserved, but not in set
5. list and set are mutable (we can perform insert, update or delete) whereas tuple are immutable

```python
#1. list have [], tuple have (), set have {}
marks = [90,89,45]  #list
ages  = (23, 18, 65) #tuple
cities= {"pune", "mumbai"} #set
print(marks, type(marks))
print(ages, type(ages))
print(cities, type(cities))
```

```python
# in list and tuple, duplicate elements are allowed but not in set
marks = [90,89,45, 90]  #list
ages  = (23, 18, 65,18) #tuple
cities= {"pune", "mumbai", "thane","pune"} #set
print(marks, type(marks))
print(ages, type(ages))
print(cities, type(cities))
```

```python
# indexing is applicable only on list and tuple, not on set
marks = [90,89,45, 90]  #list
ages  = (23, 18, 65,18) #tuple
cities= {"pune", "mumbai", "thane","pune"} #set
print(marks[0])
print(ages[0])
print(cities[0]) #indexing is not allwoed
```

# list methods

```python
#add values -----> append, insert, extend
#remove values --> remove, pop, clear
#search ---------> index
#sort -----------> sort
```

```python
# Add  Values
marks = [89]
marks.append(91)
marks.append(45)
marks.append(89)
print(marks)

marks.insert(0, 99) #it will not replace existing value/element
print(marks)

newMarks = [65, 78, 80]
marks.extend(newMarks)
print(marks)


# Remove 
marks.remove(89) # element based - first occurance of 89
print(marks)

marks.pop(0) #index based
print(marks)

marks.pop() #automatically removes last value
print(marks)


print(marks.index(78))
# print(marks.index(100)) ERROR ---> 100 is not in list

copyOfMarks = marks.copy()
print(copyOfMarks)

copyOfMarks.sort()
print(copyOfMarks)

copyOfMarks.reverse()
print(copyOfMarks)
```

# set methods

```python
# add elements -------> add, update
# remove elements ----> remove, discard
# set theory methods -> union, intersetction, difference, subset
```

```python
# Add elements
ages = {34, 98}
ages.add(54)
ages.add(18)
ages.add(34)
print(ages)

newAges = {88,77,99}
ages.update(newAges)
print(ages)

#remove elemets
ages.remove(99)
print(ages)

#ages.remove(55) ERROR ---> since  55 is not in set

ages.discard(77)
print(ages)
ages.discard(55)

set1 = {4, 7, 2, 8}
set2 = {33, 99, 2, 4}

setUnion = set1.union(set2)
print(setUnion)

setDiff = set1.difference(set2)
print(setDiff) #{8, 7}

setIntersection = set1.intersection(set2)
print(setIntersection)

set1.issubset(set2)
setDiff.issubset(set1)
```

```python
#frozenset : unlike set, duplicates are not allowed in frozen set but it is immutable
#immutable : once we create frozenset, we cant perfome inset, update, delete oprations
ageSet = {23,56,78,23}
print(ageSet)
ageSet.add(15)
print(ageSet)

ageFrozenSet = frozenset({3,7,1,3})
print(ageFrozenSet)
#ageFrozenSet.add(4) #'frozenset' object has no attribute 'add'
```

```python
#dict : dictionary 
# exmaple : student = {"name":"raj", "age":24}
# defination : dictionary is used to store data in Key:value pai
```

```python
student = {"name":"raj", "age":24}
print(student)
print(student.keys())
print(student.values())
print(student.items())

print(student.get("name"))
print(student.get("phone")) #if key is not present then we will get None

print(student)
student.update({"name":"pratik"}) #update value for existing key
print(student)

student.update({"city":"pune"}) #add new k-v pair if given is not present 
print(student)

student.setdefault("city")
student.setdefault("city","delhi") #if key is already present, do nothing
print(student)
student.setdefault("phone","98987567656") #if key is not present, add these key value pair
print(student)

student.pop("phone")
print(student)
```
